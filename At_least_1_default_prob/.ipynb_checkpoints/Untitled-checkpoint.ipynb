```python
import pandas as pd
import numpy as np
import math
```

```python
def prob_at_least_one_default(n_clients: int, p_default=0.02) -> float:
    prob_not_default = 1-p_default
    all_not_default = pow(prob_not_default,n_clients)
    at_least_one_default = 1 - all_not_default
    return at_least_one_default
```
